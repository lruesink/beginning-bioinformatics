# **<font color=blue>P3. </font>**```=``` is the ASSIGNMENT operator (assigns values to variables)
####**TASK P3.1:** Ask Gemini, "*Give me 3 examples to help practice understanding the assignment operator.*"

### Example 1: Simple Assignment
```python
# Assign the value 10 to a variable named 'number'
number = 10
print(number) # Output: 10
```
Here, the `=` operator assigns the integer value `10` to the variable `number`.

### Example 2: Re-assigning a Value
```python
# Assign a string to a variable named 'greeting'
greeting = "Hello"
print(greeting) # Output: Hello

# Re-assign a different string to the same variable
greeting = "Hi there!"
print(greeting) # Output: Hi there!
```
This example shows that you can change the value stored in a variable by re-assigning it using the `=` operator. The old value is overwritten.

### Example 3: Assigning an Expression Result
```python
# Assign the result of an arithmetic expression to 'sum_result'
num1 = 5
num2 = 7
sum_result = num1 + num2
print(sum_result) # Output: 12
```
In this case, the `=` operator assigns the outcome of the `num1 + num2` expression (which is `12`) to the variable `sum_result`.

# **<font color=blue>P4. </font>** Data types: use ```type()``` to find a variable's "type"
####**TASK P4.1:** Ask Gemini, "*Give me examples of each data type to see how they look as outputs.*"

````python
# Integer
my_integer = 10
print(f"Integer: {my_integer}, Type: {type(my_integer)}")

# Float
my_float = 3.14
print(f"Float: {my_float}, Type: {type(my_float)}")

# String
my_string = "Hello, Python!"
print(f"String: {my_string}, Type: {type(my_string)}")

# Boolean
my_boolean = True
print(f"Boolean: {my_boolean}, Type: {type(my_boolean)}")

# List
my_list = [1, 2, 3, "four"]
print(f"List: {my_list}, Type: {type(my_list)}")

# Tuple
my_tuple = (10, 20, "thirty")
print(f"Tuple: {my_tuple}, Type: {type(my_tuple)}")

# Dictionary
my_dict = {"name": "Alice", "age": 30}
print(f"Dictionary: {my_dict}, Type: {type(my_dict)}")

# Set
my_set = {1, 2, 3, 2}
print(f"Set: {my_set}, Type: {type(my_set)}")

# NoneType
my_none = None
print(f"NoneType: {my_none}, Type: {type(my_none)}")
````
